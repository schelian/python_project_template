```python
import pandas as pd
import numpy as np
import os
from pathlib import Path
```

```python
TRAIN_DIR = '../../data/CODaN/train'
VAL_DIR = '../../data/CODaN/val'
TEST_DIR = '../../data/CODaN/test'
```

```python
def images_to_csv_file(image_dir, csv_path, recursive=False, include_dir=False, extensions=None):
	"""
	Scan `image_dir` for image files and write a CSV with two columns:
	- filename: the image filename (or full path if `include_dir`)
	- class: 1 if filename contains 'day', 2 if filename contains 'night'

	Parameters:
	- image_dir (str or Path): directory to scan for images
	- csv_path (str or Path): output CSV file path
	- recursive (bool): whether to walk directories recursively
	- include_dir (bool): if True, write full file paths in the CSV; otherwise just the basename
	- extensions (iterable): allowed file extensions (e.g. ['.jpg', '.png'])

	Returns:
	- pandas.DataFrame written to `csv_path` and returned
	"""
	print( f"Processing image_dir {image_dir} and writing to {csv_path}")
 
	if extensions is None:
		extensions = {'.jpg', '.jpeg', '.png', '.bmp', '.gif', '.tif', '.tiff'}

	p = Path(image_dir)
	if recursive:
		candidates = p.rglob('*')
	else:
		candidates = p.iterdir()

	rows = []
	for f in candidates:
		if not Path(f).is_file():
			continue
		suffix = Path(f).suffix.lower()
		if suffix not in extensions:
			continue
		name = str(f) if include_dir else Path(f).name
		lname = Path(f).name.lower()
		if 'day' in lname:
			label = 1
		elif 'night' in lname:
			label = 2
		else:
			# skip files that don't match either class
			continue
		rows.append((name, label))

	df_out = pd.DataFrame(rows, columns=['filename', 'class'])
	df_out.to_csv(str(csv_path), index=False)
	return df_out
```

```python
# train index
images_to_csv_file(image_dir=TRAIN_DIR, csv_path=TRAIN_DIR + '/train.csv', recursive=False, include_dir=True, extensions=None)

# val index
images_to_csv_file(image_dir=VAL_DIR, csv_path=VAL_DIR + '/val.csv', recursive=False, include_dir=True, extensions=None)

# test index
images_to_csv_file(image_dir=TEST_DIR, csv_path=TEST_DIR + '/test.csv', recursive=False, include_dir=True, extensions=None)
```
