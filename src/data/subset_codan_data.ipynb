```python
# select subset of images
import os, random, shutil

# number of train, test points
NUM_SAMPLES_PER_CLASS = 4
NUM_TIMES_OF_DAY = 2
NUM_CLASSES = 10

VAL_SUBSET = .25
TE_SUBSET = .25

# directories
DAY_DIR = '../../data/CODaN/day/'
NIGHT_DIR = '../../data/CODaN/night/'

TRAIN_DIR = '../../data/CODaN/train'
VAL_DIR = '../../data/CODaN/val'
TEST_DIR = '../../data/CODaN/test'

random.seed(42)
```

```python
# derived constants
NUM_TR = (int)(NUM_SAMPLES_PER_CLASS*NUM_TIMES_OF_DAY*NUM_CLASSES)
NUM_TR_PER_CLASS = (int)(NUM_TR/(NUM_CLASSES*NUM_TIMES_OF_DAY))

NUM_VAL = (int)(VAL_SUBSET*NUM_TR)
NUM_VAL_PER_CLASS = (int)(NUM_VAL/(NUM_CLASSES*NUM_TIMES_OF_DAY))

NUM_TE = (int)(TE_SUBSET*NUM_TR)
NUM_TE_PER_CLASS = (int)(NUM_TE/(NUM_CLASSES*NUM_TIMES_OF_DAY))

print(f"NUM_TR: {NUM_TR}, NUM_VAL: {NUM_VAL}, NUM_TE: {NUM_TE}")
print(f"NUM_TR_PER_CLASS: {NUM_TR_PER_CLASS}, NUM_VAL_PER_CLASS: {NUM_VAL_PER_CLASS}, NUM_TE_PER_CLASS: {NUM_TE_PER_CLASS}")
```

```python
def copy_random_images_from_subfolders(base, prefix, out, num_files=4):
    print(f"source: {base}, dest: {out}, num_files: {num_files}")
    selected = {}
    for root, dirs, files in os.walk(base):
        if root == base:
            continue
        rel_path = os.path.relpath(root, base)
        prefix2 = rel_path.replace(os.sep, '_')
        image_files = [f for f in sorted(files) if os.path.isfile(os.path.join(root, f))]
        if not image_files:
            continue
        if len(image_files) < num_files:
            raise SystemExit(f"Not enough files in {root}")
        chosen = random.sample(image_files, num_files)
        selected[prefix] = chosen
        for f in chosen:
            src = os.path.join(root, f)
            dst = os.path.join(out, f"{prefix}_{prefix2}_{f}")
            shutil.copy2(src, dst)
    print('Created sample folder:', out)
    for prefix, files in selected.items():
        print(prefix + ':')
        for f in files:
            print('  ', f)
```

```python
# train from day
copy_random_images_from_subfolders(base=DAY_DIR, prefix="day", out=TRAIN_DIR, num_files=(int)(NUM_TR_PER_CLASS))
# train from night
copy_random_images_from_subfolders(base=NIGHT_DIR, prefix="night", out=TRAIN_DIR, num_files=(int)(NUM_TR_PER_CLASS))
```

```python
# val from day
copy_random_images_from_subfolders(base=DAY_DIR, prefix="day", out=VAL_DIR, num_files=(int)(NUM_VAL_PER_CLASS))
# val from night
copy_random_images_from_subfolders(base=NIGHT_DIR, prefix="night", out=VAL_DIR, num_files=(int)(NUM_VAL_PER_CLASS))
```

```python
# test from day
copy_random_images_from_subfolders(base=DAY_DIR, prefix="day", out=TEST_DIR, num_files=(int)(NUM_TE_PER_CLASS))
# test from night
copy_random_images_from_subfolders(base=NIGHT_DIR, prefix="night", out=TEST_DIR, num_files=(int)(NUM_TE_PER_CLASS))
```
